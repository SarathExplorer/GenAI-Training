```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/diabetes.csv')
df.head()
```

**2.Exploring Data set**

```python
print(df.shape,"\n")

print(df.columns,"\n")

print(df.dtypes,"\n")

print(df.isnull().sum())

df.describe()
```

**Features and Target**

```python
X= df.drop('Outcome',axis=1)
Y= df['Outcome']

print (X.shape)
print (Y.shape)

print(X.head)
print(Y.head)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score,recall_score,f1_score, confusion_matrix, classification_report,roc_auc_score,
                             roc_curve,ConfusionMatrixDisplay)
```

**Train set-Split**

```python
X_train, X_test,Y_train,Y_test = train_test_split(X,Y,test_size=0.2,random_state=42, stratify=Y)

print(X_train.shape)
print(Y_train.shape)
print(X_test.shape)
print(Y_test.shape)
```

HAndling Missing Value

```python
# Once you correct the train_test_split unpacking order in the previous cell, run this cell again:
X_train_new = X_train.replace(0, np.nan)
X_test_new = X_test.replace(0, np.nan)

print("Missing values in training set:\n", X_train_new.isnull().sum(), "\n")

imputer = SimpleImputer(missing_values=np.nan, strategy='median')
X_train_imputed = pd.DataFrame(imputer.fit_transform(X_train_new),columns = X.columns,index = X_train.index)
X_test_imputed = pd.DataFrame (imputer.transform(X_test_new),columns = X.columns,index = X_test.index)

print (pd.Series(imputer.statistics_, index = X.columns))

print("\n",X_train_imputed.isnull().sum())



```

Categorical Variable

No Categorical value columns has been identified. So no columns has been convereted.


```python
print (X.dtypes,"\n")

print (X_train.dtypes,"\n")

print (X_train_imputed.dtypes)
```

Standardisation

```python
Scaler = StandardScaler()
X_train_scaled = pd.DataFrame(Scaler.fit_transform(X_train_imputed),columns = X_train_imputed.columns,index = X_train_imputed.index)
X_test_scaled = pd.DataFrame(Scaler.transform(X_test_imputed),columns = X_test_imputed.columns,index = X_test_imputed.index)

print (X_train_imputed.head())

print(X_train_scaled.head())
```

Machine Learning Model Building

Logistic regression is being used for the Model building due to the follwing parameters:
1. Binary Target : Outcome is either 0 or 1
2. We will get effect of each feature on the outcome using the coefficient after prediction.
3. Fast and consistently providing 70-80% accuracy.

```python
lr = LogisticRegression(random_state=42)

print (lr)
```

Model Training

```python
lr.fit(X_train_scaled,Y_train)
print ("Intercept",lr.intercept_)
print("iterations", lr.n_iter_)

coef_df = pd.DataFrame({'Features':X_train_scaled.columns,'Coefficients':lr.coef_[0]})
print(coef_df)
```

Prediction

```python
Y_train_Pre = lr.predict(X_train_scaled)
Y_train_pro = lr.predict_proba(X_train_scaled)[:,1]

Output = pd.DataFrame({'Actual':Y_train,'Predicted':Y_train_Pre,'Probability':Y_train_pro})
Output.head(50)
```

Performance

```python
Pred_Accu = accuracy_score(Y_train,Y_train_Pre)
print(f"Accuracy Score{Pred_Accu: .4%}")
```

Evaluate and Predict Test Data

```python
Y_test_Pre = lr.predict(X_test_scaled)
Y_test_pro = lr.predict_proba(X_test_scaled)[:,1]

test_Output = pd.DataFrame({'Actual':Y_test,'Predicted':Y_test_Pre,'Probability':Y_test_pro})
test_Output.head(50)
```

```python
test_accu = accuracy_score(Y_test,Y_test_Pre)
print(f"Accuracy Score{test_accu: .4%}")

test_prec = precision_score(Y_test,Y_test_Pre)
print(f"Precision Score{test_prec: .4%}")

test_recall = recall_score(Y_test,Y_test_Pre)
print(f"Recall Score{test_recall: .4%}")

test_f1 = f1_score(Y_test,Y_test_Pre)
print(f"F1 Score{test_f1: .4%}")
```

Confusion Matrix

```python
Con_Mat = confusion_matrix(Y_test,Y_test_Pre)
True_Neg, False_Pos, False_Neg, True_Pos = Con_Mat.ravel()
print(f"True Negative: {True_Neg}, False Positive: {False_Pos}, False Negative: {False_Neg}, True Positive: {True_Pos}")
```

Comparison of Training and Testing Performance

```python
def all_mat (Y_true,Y_Pre):
  return{
      "Training Accuracy": accuracy_score(Y_true,Y_Pre),
      "Training Precision": precision_score(Y_true,Y_Pre),
      "Training Recall": recall_score(Y_true,Y_Pre),
      "Training F1": f1_score(Y_true,Y_Pre)
  }
Comparison = pd.DataFrame({"Training":all_mat(Y_train,Y_train_Pre),"Testing":all_mat(Y_test,Y_test_Pre)})
Comparison["Gap"] = Comparison["Training"] - Comparison["Testing"]
Comparison
```

***Conclusions***

The Model perforance is statisfactory providing 70% of accuracy in testing data. However there is observable difference of 9 points between the accuracy of the Trainig and test data.

The current model is able to provide us the dependency of input features based on the coefficient.

The gap in the acuuracy and precision between the Training
and testing we could assume there could be slight overfitting.
